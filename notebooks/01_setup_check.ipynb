```python
from PIL import Image
from pathlib import Path

content = Image.open("../data/content/Abyssinian_24.jpg").convert("RGB")
style = Image.open("../data/style/1280px-The_Great_Wave_off_Kanagawa.jpg").convert("RGB")
```

```python
from PIL import ImageOps

resized_content = ImageOps.contain(content, (384, 384), method=Image.Resampling.LANCZOS)
resized_style = ImageOps.contain(style, (384, 384), method=Image.Resampling.LANCZOS)
```

```python
resized_content.size
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1,2, figsize=(12,5))
axes[0].imshow(resized_content)
axes[0].set_title("Content_image")
axes[0].axis("off")
axes[1].imshow(resized_style)
axes[1].set_title("Style_image")
axes[1].axis("off")
```
